````python
a = np.array([1, 2])
b = np.array([2, 1])

cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle = np.degrees(np.arccos(cos_angle))

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

A_normalized = (A - mean) / std

print("Середнє по стовпцях:", mean)
print("Стандартне відхилення:", std)
print("Нормалізована матриця:")
print(A_normalized)
````

````python
sales = np.array([
    [120, 150, 170, 160],
    [80, 90, 100, 110],
    [200, 180, 210, 230]
])

product_totals = np.sum(sales, axis=1)
month_totals = np.sum(sales, axis=0)

print("Продажі кожного товару:", product_totals)
print("Продажі у кожному місяці:", month_totals)
print("Товар з найбільшими продажами:", np.argmax(product_totals) + 1)
print("Місяць з найбільшими продажами:", np.argmax(month_totals) + 1)
print("Середні продажі кожного товару:", np.mean(sales, axis=1))
````

Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
np.random.seed(42)

A = np.random.randint(1, 21, (5, 5))

main_diagonal_sum = np.trace(A)
secondary_diagonal_sum = np.trace(np.fliplr(A))
row_max = np.max(A, axis=1)
column_min = np.min(A, axis=0)
mean = np.mean(A)

A_changed = A.copy()
A_changed[A_changed < mean] = 0

print("Матриця:")
print(A)
print("Сума головної діагоналі:", main_diagonal_sum)
print("Сума побічної діагоналі:", secondary_diagonal_sum)
print("Максимум кожного рядка:", row_max)
print("Мінімум кожного стовпця:", column_min)
print("Середнє значення:", mean)
print("Матриця після заміни:")
print(A_changed)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
A = np.array([
    [1, 2, 3, 4],
    [2, 3, 4, 5],
    [5, 6, 7, 8],
    [1, 3, 5, 7]
])

norms = np.linalg.norm(A, axis=1, keepdims=True)
A_normalized = A / norms

print("Норми рядків:")
print(norms)

print("Нормалізована матриця:")
print(A_normalized)

print("Перевірка норм:")
print(np.linalg.norm(A_normalized, axis=1))
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])
b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)

print("Розв'язок [x, y, z]:", x)
print("A @ x:", A @ x)
print("b:", b)
print("Результати збігаються:", np.allclose(A @ x, b))
````
